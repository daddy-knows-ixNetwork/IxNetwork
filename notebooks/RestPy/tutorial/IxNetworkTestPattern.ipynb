# IxNetwork Test Pattern

## 5-phase life-cycle

* Setup
    - Start a new session
    - Map, Assign and Connect Ports
* Configuration
    - Create Topology
    - Create Device Groups (Protocol Stacks)
    - Create Network Groups (Network Stacks) 
* Execution (make sure the synchronization)
    - Start Protocols
    - Start Traffic
    - Stop Protocols
    - Stop Traffic
* Verification
    - Analyze/Export results to xlsx, csv, etc.
* Teardown
    - Unmap, Unassign and Disconnect Ports
    - Remove Session
